```python
roll_no=input("Enter your 10-digit roll number: ")
L=[int(d)*10 for d in roll_no]
print("L:",L)

L.append(100)
print("After append:",L) # append adds 100 at the end

L.insert(2,200)
print("After insert:",L) # insert adds 200 at index 2

L.remove(100)
print("After remove:",L)

removed=L.pop(2)
print("After pop:",L)

L.sort()
print("Ascending:",L)

L.sort(reverse=True)
print("Descending:",L)

print("First three:",L[:3])
print("Last three:",L[-3:])

average=sum(L)/len(L)
new_list=[x for x in L if x>average]
print("Average:",average)
print("Greater than average:",new_list)
```

```python
scores=tuple(L[:8])
print("Scores:",scores)

highest=max(scores)
print("Highest:",highest)
print("Highest index:",scores.index(highest))

lowest=min(scores)
print("Lowest:",lowest)
print("Lowest count:",scores.count(lowest))

reversed_scores=list(reversed(scores))
print("Reversed list:",reversed_scores) # Tuples are immutable, so they cannot be reversed in place

user_score=int(input("Enter a score: "))

if user_score in scores:
    print("First occurrence index:",scores.index(user_score))
else:
    print("Score is not present")

try:
    scores[0]=100
except TypeError as e:
    print("Error:",e)
    print("Tuples are immutable, so their elements cannot be changed unlike lists.")

first_score,second_score,*remaining_scores=scores
print("First score:",first_score)
print("Second score:",second_score)
print("Remaining scores:",remaining_scores)
```

```python
import random

random.seed(int(roll_no))

numbers=[random.randint(100,900) for _ in range(100)]
print("Numbers:",numbers)

odd_numbers=[x for x in numbers if x%2!=0]
print("Odd numbers:",odd_numbers)
print("Odd count:",len(odd_numbers))

even_numbers=[x for x in numbers if x%2==0]
print("Even numbers:",even_numbers)
print("Even count:",len(even_numbers))

def prime(n):
    if n<2:
        return False
    for i in range(2,int(n**0.5)+1):
        if n%i==0:
            return False
    return True

prime_numbers=[x for x in numbers if prime(x)]
print("Prime numbers:",prime_numbers)
print("Prime count:",len(prime_numbers))

most_frequent=max(set(numbers),key=numbers.count)
print("Most frequent number:",most_frequent)
print("Frequency:",numbers.count(most_frequent))
```

```python
digits=[int(d) for d in roll_no[:8]]

A={d*7 for d in digits}
B={d*9 for d in digits}

print("A:",A)
print("B:",B)

print("Union:",A.union(B))
print("Intersection:",A.intersection(B))

print("A-B:",A.difference(B))
print("B-A:",B.difference(A))
print("difference() gives elements in one set but not the other, while symmetric_difference() gives elements in either set but not both.")

print("Symmetric difference:",A.symmetric_difference(B))

print("A subset of B:",A.issubset(B))
print("B superset of A:",B.issuperset(A))

X=int(input("Enter a value to remove from A: "))
A.discard(X)
print("A after discard:",A)
print("discard() is safer because it does not give an error if the value is not present.")
```

```python
name=input("Enter your name: ")
branch=input("Enter your branch: ")
age=int(input("Enter your age: "))
city=input("Enter your home city: ")
cgpa=float(input("Enter your CGPA: "))

my_dict={"name":name,"roll_no":roll_no,"branch":branch,"age":age,"city":city}

print("Original:",my_dict)

my_dict["location"]=my_dict.pop("city")
my_dict["cgpa"]=cgpa
print("After changes:",my_dict)

my_dict["age"]=my_dict["age"]+1
print("After increasing age:",my_dict)

dict1=my_dict.copy()
dict2=my_dict.copy()

removed=dict1.pop("branch")
del dict2["branch"]

print("Using pop():",dict1)
print("Removed value:",removed)
print("Using del:",dict2)
print("pop() returns the deleted value, while del only deletes it.")

for key,value in my_dict.items():
    print(key,"→",value)

if "email" in my_dict:
    print("Email:",my_dict["email"])
else:
    print("Email does not exist.")

friend_dict={"name":"Rahul","roll_no":"12345678","branch":"CSE","age":20,"city":"Delhi"}

merged={**my_dict,**friend_dict}
print("Friend dictionary:",friend_dict)
print("Merged dictionary:",merged)
print("When both dictionaries have the same key, the value from the second dictionary wins.")

string_dict={key:value for key,value in my_dict.items() if isinstance(value,str)}
print("String values:",string_dict)
```
