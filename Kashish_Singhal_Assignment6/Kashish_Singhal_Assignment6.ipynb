## Q1. Sensor Readings - Vectorization and Boolean Indexing

```python
temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])
print("Original readings:", temperature)
```

```python
corrected = temperature + 2
print("a) After adding 2 degree correction:", corrected)
```

```python
fahrenheit = (9/5) * temperature + 32
print("b) Celsius to Fahrenheit:", fahrenheit)
```

```python
above_32 = temperature[temperature > 32]
print("c) Readings greater than 32C:", above_32)
```

```python
count = np.sum(temperature > 32)
print("d) Number of readings exceeding 32C:", count)
```

**e) Vectorization vs Looping:**

With a regular for loop, Python processes one element at a time which means a lot of overhead for each iteration. Vectorization in NumPy applies the operation to the entire array at once using optimized C code internally, so it's much faster especially as array size grows.

Boolean indexing works the same way — instead of checking each element with an if statement inside a loop, NumPy creates a boolean mask for the whole array in one go and uses it to filter values. For large datasets like real sensor streams, this difference in speed is very noticeable.

## Q2. Daily Steps Matrix - Aggregation and Axis Operations

```python
steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])
print("Steps matrix:")
print(steps)
```

```python
print("a) Total steps recorded:", np.sum(steps))
```

```python
print("b) Mean steps:", np.mean(steps))
```

```python
print("c) Maximum value:", np.max(steps))
print("   Minimum value:", np.min(steps))
```

```python
print("d) Total steps per day (axis=0):", np.sum(steps, axis=0))
```

```python
print("e) Total steps per user (axis=1):", np.sum(steps, axis=1))
```

```python
flat_idx = np.argmax(steps)
row, col = np.unravel_index(flat_idx, steps.shape)
print("f) Position of maximum steps -> User (row):", row, ", Day (col):", col)
print("   Value at that position:", steps[row, col])
```

## Q3. Slicing, Copying, Reshaping and Flattening

```python
original = np.array([1, 2, 3, 4, 5, 6])
print("a) Original array:", original)
```

```python
subset = original[1:5]
print("b) Subset (index 1 to 4):", subset)
```

```python
subset[0] = 999
print("c) After modifying subset[0] to 999:")
print("   subset:", subset)
print("   original:", original)
print("   (original also changed because subset is a view, not a copy)")
```

```python
original = np.array([1, 2, 3, 4, 5, 6])
copied = original[1:5].copy()
copied[0] = 500
print("d) After modifying copied[0] to 500:")
print("   copied array:", copied)
print("   original array:", original)
print("   (original is unchanged because .copy() creates an independent array)")
```

```python
matrix = np.arange(1, 13).reshape(3, 4)
print("e) 3x4 matrix using arange and reshape:")
print(matrix)
```

```python
print("f) Slicing operations:")
print("   First row:", matrix[0, :])
print("   Last row:", matrix[-1, :])
print("   Second column:", matrix[:, 1])
print("   Rows 1-2, Columns 2-3:")
print(matrix[1:3, 2:4])
```

```python
flat_flatten = matrix.flatten()
flat_ravel = matrix.ravel()
print("g) flatten():", flat_flatten)
print("   ravel(): ", flat_ravel)
```

```python
flat_ravel[0] = 999
print("h) After modifying ravel()[0] to 999:")
print("   ravel array:", flat_ravel)
print("   original matrix:")
print(matrix)
print("   (ravel() returns a view so the original matrix is also changed)")
```

```python
matrix = np.arange(1, 13).reshape(3, 4)
flat_flatten = matrix.flatten()
flat_flatten[0] = 500
print("i) After modifying flatten()[0] to 500:")
print("   flattened array:", flat_flatten)
print("   original matrix:")
print(matrix)
print("   (flatten() returns a copy so the original is not affected)")
```

```python
matrix = np.arange(1, 13).reshape(3, 4)
print("j) Properties of the 3x4 matrix:")
print("   shape:", matrix.shape)
print("   ndim:", matrix.ndim)
print("   size:", matrix.size)
print("   dtype:", matrix.dtype)
```

## Q4. Cognitive Assistive System - OLS Regression using NumPy

```python
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])

y = np.array([40, 65, 30, 85])

print("a) Shape of X:", X.shape)
print("   Dimensions of X:", X.ndim)
```

```python
print("b) Transpose of X (X.T):")
print(X.T)
print()
print("The transpose flips rows and columns. Original X has shape (4,3) meaning 4 users x 3 features.")
print("X.T has shape (3,4) — each row now represents one feature across all 4 users.")
print("This is needed for matrix multiplication in OLS.")
```

```python
XtX = X.T @ X
print("c) X.T @ X:")
print(XtX)
```

```python
XtX_inv = np.linalg.inv(XtX)
print("d) Inverse of (X.T @ X):")
print(XtX_inv)
```

```python
beta = XtX_inv @ X.T @ y
print("e) OLS coefficients (beta = (X.T X)^-1 X.T y):")
print(beta)
```

**f) What the coefficients mean:**

Each coefficient in `beta` tells us how much the assistance score changes for a one-unit increase in that feature, keeping the other two constant.

- **beta[0]** → effect of Sleep hours on assistance score
- **beta[1]** → effect of Activity level on assistance score  
- **beta[2]** → effect of Stress level on assistance score

A negative coefficient for sleep/activity would mean higher values lower the score (less assistance needed), and a positive coefficient for stress would mean higher stress raises the score (more assistance needed). The exact direction depends on the computed values.

```python
new_user = np.array([5, 40, 7])
predicted_score = new_user @ beta
print("g) New user features:", new_user)
print("   Predicted assistance score:", round(predicted_score, 2))
```
