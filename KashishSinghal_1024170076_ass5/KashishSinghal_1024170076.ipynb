## Q2. Basic NumPy Array Operations

```python
arr = np.array([1, 2, 3, 6, 4, 5])
print("a) Reversed array:", arr[::-1])
```

```python
x = np.array([1,2,3,4,5,1,2,1,1,1])

vals, counts = np.unique(x, return_counts=True)
most_freq = vals[np.argmax(counts)]
indices = np.where(x == most_freq)[0]

print("b i) Array x:", x)
print("Most frequent value:", most_freq)
print("Indices:", indices)
```

```python
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

vals, counts = np.unique(y, return_counts=True)
most_freq = vals[np.argmax(counts)]
indices = np.where(y == most_freq)[0]

print("b ii) Array y:", y)
print("Most frequent value:", most_freq)
print("Indices:", indices)
```

## Q4. linspace() - Array Properties and Transpose

```python
Kashish Singhal = np.linspace(10, 100, 25)
print("Array:", Kashish Singhal)
```

```python
print("Dimensions (ndim):", Kashish Singhal.ndim)
print("Shape:", Kashish Singhal.shape)
print("Total elements (size):", Kashish Singhal.size)
print("Data type:", Kashish Singhal.dtype)
print("Total bytes:", Kashish Singhal.nbytes)
```

```python
transposed_reshape = Kashish Singhal.reshape(25, 1)
print("Transpose using reshape():")
print(transposed_reshape)
```

```python
print("Transpose using .T attribute:")
print(Kashish Singhal.T)
print()
print("For a 1D array, .T does not change anything since there's only one dimension.")
print("We need to first reshape it to 2D to see any effect with .T")
print()
reshaped_2d = Kashish Singhal.reshape(1, 25)
print("After reshape to (1,25), transpose using .T:")
print(reshaped_2d.T)
```

## Q5. 2D Array - Statistics, Reshape, Resize

```python
ucs420_Kashish Singhal = np.array([[10, 20, 30, 40], [50, 60, 70, 80], [90, 15, 20, 35]])
print("ucs420_Kashish Singhal:")
print(ucs420_Kashish Singhal)
```

```python
print("Mean:", np.mean(ucs420_Kashish Singhal))
print("Median:", np.median(ucs420_Kashish Singhal))
print("Max:", np.max(ucs420_Kashish Singhal))
print("Min:", np.min(ucs420_Kashish Singhal))
print("Unique elements:", np.unique(ucs420_Kashish Singhal))
```
